```python
#Top Selling Products (Product popularity)
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["ecommerce_db"]

pipeline = [

    # Break items array into individual products
    {
        "$unwind": "$items"
    },

    # Group by product
    {
        "$group": {
            "_id": "$items.product_id",

            "total_quantity_sold": {
                "$sum": "$items.quantity"
            },

            "total_revenue": {
                "$sum": "$items.subtotal"
            },

            "number_of_orders": {
                "$sum": 1
            }
        }
    },

    # Sort descending
    {
        "$sort": {
            "total_quantity_sold": -1
        }
    },

    # Top 10 products
    {
        "$limit": 10
    }
]

results = db.transactions.aggregate(pipeline)

print("\nTOP 10 SELLING PRODUCTS\n")

for r in results:
    print(
        f"Product: {r['_id']}"
        f"\nQuantity Sold: {r['total_quantity_sold']}"
        f"\nRevenue: ${r['total_revenue']:,.2f}"
        f"\nOrders: {r['number_of_orders']}\n"
    )
```

```python
#Revenue by Payment Method
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
db = client["ecommerce_db"]

pipeline = [
    {
        "$group": {
            "_id": "$payment_method",
            "total_revenue": {"$sum": "$total"},
            "transaction_count": {"$sum": 1},
            "average_order_value": {"$avg": "$total"}
        }
    },
    {
        "$sort": {
            "total_revenue": -1
        }
    }
]

results = db.transactions.aggregate(pipeline)

print("\n===== REVENUE BY PAYMENT METHOD =====\n")

for r in results:
    print(
        f"Payment Method: {r['_id']}\n"
        f"Revenue: ${r['total_revenue']:,.2f}\n"
        f"Transactions: {r['transaction_count']}\n"
        f"Average Order Value: ${r['average_order_value']:,.2f}\n"
    )
```
